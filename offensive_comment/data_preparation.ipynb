```python
import pandas as pd
import spacy 
```

```python
data = pd.read_csv('offensive_comment_small.csv', sep=';')
data.info()
```

```python
data.head()
```

```python
nlp = spacy.load("en_core_web_sm", disable=["ner"])

def clean_text(text):
  tab_result = []
  # Ne garder que les LEMME, enlever les STOPWORDS et les mots de moins de 2 charactères
  doc = nlp(text)
  for token in doc :
    if token.is_stop == False and len(token.text)>2:
      tab_result.append(token.lemma_)
  # A vous de coder
  return tab_result

# modifier la phrase
clean_text("This IS a test, we are currently in a big data lesson and we try spacy")
```

```python
data['clean_comment'] = data['comment_text'].apply(clean_text)
data.head()
```
